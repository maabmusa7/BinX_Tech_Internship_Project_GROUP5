# English Tutor LLM

## Project Goal

The goal of this project is to fine-tune a Large Language Model (LLM) to act as an English conversation tutor.

The model should be able to:

* Hold natural English conversations with the student.
* Adapt its language to the student's CEFR level.
* Detect important grammar mistakes.
* Correct mistakes and provide simple explanations.
* Ask follow-up questions to continue the conversation.
* Maintain the context of multi-turn conversations.


## 1-Base Model

**Model:** Qwen3 4B

**Framework:** TensorFlow / Keras

**Library:** KerasHub

### Why We Chose This Model

We chose Qwen3 4B because it provides a strong balance between model capability and computational requirements.

It is suitable for conversational and instruction-following tasks, while its 4-billion-parameter size makes it more practical for experimentation and fine-tuning than significantly larger language models.

Using KerasHub also allows us to work with the same TensorFlow/Keras ecosystem used throughout the project.
