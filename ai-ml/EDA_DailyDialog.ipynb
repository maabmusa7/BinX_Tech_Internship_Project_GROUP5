## Exploratory Data Analysis — DailyDialog

The purpose of this EDA is to understand the structure and characteristics of the DailyDialog dataset before preprocessing and fine-tuning.


### Convert the Dataset to a Pandas DataFrame

The DailyDialog dataset is converted into a Pandas DataFrame to make data exploration and analysis easier.

```python
from datasets import load_dataset
daily_dialog = load_dataset("elricwan/dailydialog")
df = daily_dialog["train"].to_pandas()
df.head()
```

### Dataset Shape

The dataset shape shows the number of conversations and the number of available columns.

```python
df.shape
```

### Dataset Information

We inspect the dataset structure, column names, data types, and number of non-null values.

```python
df.info()
```

### Empty Conversations

A conversation may exist in the dataset without being null but still contain no dialogue turns. Therefore, empty conversations are checked separately.

```python
empty_conversations = df["conversation"].apply(lambda x: len(x) == 0).sum()
print("Empty conversations:", empty_conversations)
```
